```python
class Animal:
    registry = []   # just a plain list

    def __init_subclass__(cls):
        # Python calls this automatically whenever
        # someone does class Dog(Animal)
        Animal.registry.append(cls)

class Dog(Animal):
    pass

class Cat(Animal):
    pass

class Snake(Animal):
    pass

print(Animal.registry)
# [<class 'Dog'>, <class 'Cat'>, <class 'Snake'>]
```

```python
class AnimalMeta(type):
    """
    This is the metaclass.
    It runs every time a new class is DEFINED — not instantiated.
    Think of it as a class factory with a side effect.
    """
    registry = {}   # tablename -> class

    def __new__(mcs, name, bases, namespace):
        # This fires the moment Python reads: class Dog(Animal): ...
        cls = super().__new__(mcs, name, bases, namespace)

        # Don't register the base Animal class itself
        if bases:
            table = namespace.get("__tablename__")
            if table:
                AnimalMeta.registry[table] = cls
                print(f"  Registered: '{table}' -> {cls.__name__}")

        return cls


class Animal(metaclass=AnimalMeta):
    """Base class. Just by using metaclass=AnimalMeta,
    every subclass automatically goes through AnimalMeta.__new__"""
    pass


print("Defining Dog...")
class Dog(Animal):
    __tablename__ = "dogs"
    sound = "woof"

print("Defining Cat...")
class Cat(Animal):
    __tablename__ = "cats"
    sound = "meow"

print("Defining Snake...")
class Snake(Animal):
    __tablename__ = "snakes"
    sound = "hiss"

print()
print("Registry now contains:")
print(AnimalMeta.registry)
for table, cls in AnimalMeta.registry.items():
    print(f"  '{table}' -> {cls.__name__}")
```
